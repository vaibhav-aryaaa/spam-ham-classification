```python
lemmatizer = WordNetLemmatizer()
```

```python
import nltk,re
nltk.download('stopwords')
from nltk.corpus import stopwords
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
corpus=[]
for i in range(0,len(messages)):
    review=re.sub('[^a-zA-z]',' ',messages['message'][i]) #substituiting, apart from a-z and A-Z, if you get any special character then just replace it with the blank and try to replace it in our message and store it in our review variable
    review=review.lower()
    review=review.split()
    review=[lemmatizer.lemmatize(word) for word in review]
    review=' '.join(review)
    corpus.append(review)
```
